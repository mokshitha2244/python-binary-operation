```python
def to_binary(num):
    return format(num & 255, '08b')


def add_numbers(a, b):
    binary_a = to_binary(a)
    binary_b = to_binary(b)

    result = a + b
    binary_result = to_binary(result)

    print("First number:", a, "=", binary_a)
    print("Second number:", b, "=", binary_b)
    print("Binary addition:", binary_result)

    # Check whether the result is within the 8-bit signed range
    if result < -128 or result > 127:
        print("Overflow occurred!")
    else:
        print("Decimal result:", result)


print("a) 25 + (+18)")
add_numbers(25, 18)

print("\nb) (-25) + (-18)")
add_numbers(-25, -18)
```
