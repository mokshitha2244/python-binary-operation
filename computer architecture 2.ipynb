```python
#left shift
num = int("10110110", 2)

print("Logical Left Shift")
print("Original number:", format(num, "08b"))

for i in range(2):
    num = (num << 1) & 255
    print("After shift", i + 1, ":", format(num, "08b"))

print("Final result:", format(num, "08b"))


# Logical right shift
num = int("11010100", 2)

print("\nLogical Right Shift")
print("Original number:", format(num, "08b"))

for i in range(2):
    num = num >> 1
    print("After shift", i + 1, ":", format(num, "08b"))

print("Final result:", format(num, "08b"))
```
